```python
def categorize_features(sample):
    wind, vis, age, cong = sample

    # 1. Categorize Wind Speed
    if wind <= 10:
        cat_wind = 1
    elif wind <= 20:
        cat_wind = 2
    elif wind <= 30:
        cat_wind = 3
    else:
        cat_wind = 4

    # 2. Categorize Visibility
    if vis < 3.0:
        cat_vis = 1
    elif vis <= 7.0:
        cat_vis = 2
    else:
        cat_vis = 3

    # 3. Categorize Plane Age
    if age <= 5:
        cat_age = 1
    elif age <= 15:
        cat_age = 2
    else:
        cat_age = 3

    # 4. Categorize Airport Congestion
    if cong <= 30:
        cat_cong = 1
    elif cong <= 60:
        cat_cong = 2
    elif cong <= 80:
        cat_cong = 3
    else:
        cat_cong = 4

    return [cat_wind, cat_vis, cat_age, cat_cong]

X_train_cat = [categorize_features(x) for x in X_train]
```

```python
import numpy as np

def distance(a,b):
  dist = 0
  for i in range(len(a)):
    dist += (a[i]-b[i])**2
  dist = dist**0.5
  return dist
def knn(data, labels, x, N):
  arr = np.array([distance(i,x) for i in data])
  indices_unsorted = np.argpartition(arr, N)[:N]
  y = sum([labels[i] for i in indices_unsorted])/N
  return 0 if y<=0.5 else 1

testing = [knn(X_train_cat, y_train , categorize_features(i), 3) for i in X_test]
print(testing, y_test)
```
